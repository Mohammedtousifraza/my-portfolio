Step 1: Add Guardrails to Static Context

```python
SYSTEM_CONTEXT = """
You are a Hospital Appointment Assistant.

Guardrails:
- Be polite and professional.
- Help users book, reschedule, or cancel appointments.
- Do not provide medical diagnosis or treatment advice.
- Do not invent doctors, appointment slots, dates, or hospital information.
- Use only the information provided in the context.
- If required information is missing, ask the user for it.
- For emergency situations, advise the user to contact the emergency department.
- Protect patient privacy and do not request unnecessary personal information.
"""
```

Step 2: Incomplete User Context

```python
user_query = "I want to book an appointment with a doctor tomorrow morning."

user_profile = {
    "patient_name": "Mohammed",
    "preferred_time": "Morning"
}

appointment_context = {
    "date": "15 September 2026",
    "department": None,
    "doctor": None,
    "available_slots": None
}
```

Step 3: Assemble Context with Missing Information

```python
final_prompt_with_missing_info = f"""
{SYSTEM_CONTEXT}

Session Memory:
Patient Name: {user_profile["patient_name"]}
Preferred Time: {user_profile["preferred_time"]}

Appointment Information:
Date: {appointment_context["date"]}
Department: {appointment_context["department"]}
Doctor: {appointment_context["doctor"]}
Available Slots: {appointment_context["available_slots"]}

User Question:
{user_query}

Task:
Identify the missing information required to book the appointment.
Do not guess the missing information.
Ask the patient for the required details.
"""
```

```python
from google.colab import userdata
from openai import OpenAI

# 1. Load API key from Colab Secrets
api_key = userdata.get("api_key")

# 2. Create OpenAI-compatible client
client = OpenAI(
    api_key=api_key,
    base_url="https://nexusapi.navigatelabs.ai"
)

# 3. Send request to the model
response = client.chat.completions.create(
    model="gemini-2.5-flash",
    messages=[
        {
            "role": "system",
            "content": SYSTEM_CONTEXT
        },
        {
            "role": "user",
            "content": final_prompt_with_missing_info
        }
    ]
)

# 4. Print model response
print(response.choices[0].message.content)
```
